### Problem 1 — One mean, infinite population (no FPC)

Estimate **mean minutes/day** on course website; $n=25$.

**Tasks:** choose formula · mean & SD · histogram + dot plot · 95% CI.


### Code anatomy — Problem 1

| Part | Role |
|------|------|
| `<span style="color:#2563eb">mean(x)</span>` | Sample mean $\bar{x}$ |
| `<span style="color:#2563eb">t.test(x)</span>` | 95% CI for population mean (no FPC) |
| `<span style="color:#dc2626">x <- c(...)</span>` | Replace with your data in Project 2 |


````r
x <- c(12, 18, 25, 9, 31, 16, 20, 22, 14, 3, 19, 11, 24, 30, 15, 15, 21, 26, 13, 28, 10, 23, 29, 18, 33)

mean(x)
sd(x)

ggplot(data.frame(x = x), aes(x = x)) +
  geom_histogram(binwidth = 5, boundary = 0) +
  geom_dotplot(binwidth = 1, stackdir = "up", dotsize = 0.75) +
  labs(title = "Daily minutes on course website (sample)",
       x = "Minutes per day", y = "Count")

t.test(x)
````

**Read output:** <span style="color:#7c3aed">mean.conf.int</span> gives the 95% CI endpoints. Project 2 mean template uses the same pattern with your SRS data.


### Problem 2 — One mean, finite population (FPC)

$N=300$ classrooms; SRS $n=30$ without replacement; chair counts.


````r
N <- 300
x <- c(28, 30, 32, 27, 27, 33, 33, 26, 30, 30, 28, 29, 35, 27, 32, 32, 28, 30, 29, 33, 26, 34, 28, 31, 30, 27, 32, 29, 33, 28)

n <- length(x)
mean(x)
sd(x)

ggplot(data.frame(x = x), aes(x = x)) +
  geom_histogram(binwidth = 5, boundary = 0) +
  geom_dotplot(binwidth = 1, stackdir = "up", dotsize = 0.5) +
  labs(title = "Chairs per classroom (SRS from finite population)",
       x = "Number of chairs", y = "Count")

# 95% t-interval WITH FPC
mean(x) + c(-1, 1) * qt(0.975, df = n - 1) * (sd(x) / sqrt(n)) * sqrt((N - n) / (N - 1))
````

### Problem 3 — One proportion, large population (no FPC)

$n=220$, $x=80$ with part-time job.


````r
n <- 220
x <- 80
job <- c(rep(TRUE, x), rep(FALSE, n - x))

mean(job)  # p-hat

ggplot(data.frame(job = job), aes(x = job)) +
  geom_bar() +
  labs(title = "Part-time job? (sample of CA students)",
       x = "Has part-time job?", y = "Count")

prop.test(x, n, correct = FALSE)
````

### Problem 4 — One proportion, finite population (FPC)

$N=1500$ schools; $n=108$; $x=20$ older than 50 years.


````r
N <- 1500
n <- 108
x <- 20
old50 <- c(rep(TRUE, x), rep(FALSE, n - x))

p_hat <- mean(old50)
p_hat + c(-1, 1) * qnorm(0.975) * sqrt(p_hat * (1 - p_hat) / n) * sqrt((N - n) / (N - 1))
````

### Problem 5 — One proportion, small sample (exact binomial)

$n=10$, $x=3$ offer AP Stats — large-count condition fails.


````r
n <- 10
x <- 3
apstats <- c(rep(TRUE, x), rep(FALSE, n - x))

mean(apstats) * n
(1 - mean(apstats)) * n

binom.test(x, n)$conf.int
````
