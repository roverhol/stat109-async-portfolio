## Part 6 — Simulation: 4 babies returned to 4 mothers

### Story

Four mothers and four babies. Each baby is returned to a **random** mother (a random shuffle).

**Questions:**

- What is P(all 4 correct)?
- What is P(none correct)?

We will **simulate** many shuffles in R — the same idea as shuffling cards in class, but automated.


### 6.1 One trial (one shuffle)

**Why we do this:** `sample()` randomly shuffles the numbers 1–4. Each run can give a different assignment.

**Your turn next (exercise notebook):** Run the cell several times *without* `set.seed` and watch `num_matches` change.
### Code anatomy — sample()

| Part | Role |
|------|------|
| <span style="color:#2563eb">sample</span> | random shuffle |
| (<span style="color:#dc2626">c(1, 2, 3, 4)</span>) | pool to shuffle |
| <span style="color:#059669">babies</span> | result — who got which baby |
| <span style="color:#2563eb">sum</span> | count matches after `babies == mothers` |


```r
mothers <- c(1, 2, 3, 4)

babies <- sample(c(1, 2, 3, 4))
babies

num_matches <- sum(babies == mothers)
num_matches
```

**After you run the cell:** `num_matches` is 0–4. One shuffle is one **replicate** of the random process.


### 6.2 Ten trials with a `for` loop

**Why we do this:** Repeat the shuffle 10 times and store each result in a vector.

**Your turn next (exercise notebook):** In the exercise notebook you will write a similar loop.

**Tip:** `set.seed(109)` makes randomness reproducible so your demo matches ours.
### Code anatomy — for loop

| Part | Role |
|------|------|
| <span style="color:#2563eb">for</span> | repeat the block for each index i |
| (<span style="color:#dc2626">i in 1:10</span>) | how many replicates — change count here |
| <span style="color:#059669">num_correct[i]</span> | store result of replicate i |
| <span style="color:#2563eb">set.seed</span> | fix randomness for reproducibility |


```r
set.seed(109)

mothers <- c(1, 2, 3, 4)
num_correct <- numeric(10)

for (i in 1:10) {
  babies <- sample(c(1, 2, 3, 4))
  num_correct[i] <- sum(babies == mothers)
}

num_correct
```

**After you run the cell:** You get 10 numbers — one match-count per shuffle. Most runs are not 4 or 0.


### 6.3 Many trials — estimate probabilities

**Why we do this:** Run 10,000 shuffles. The **proportion** of replicates with 4 matches estimates P(all 4 correct).

**Your turn next (exercise notebook):** You will run this block in the exercise notebook.
### Code anatomy — mean on logicals

| Part | Role |
|------|------|
| <span style="color:#059669">num_correct</span> | match counts from simulation |
| <span style="color:#2563eb">==</span> | which replicates hit 4 (or 0)? |
| <span style="color:#2563eb">mean</span> | proportion TRUE ≈ probability |


```r
set.seed(109)

mothers <- c(1, 2, 3, 4)
num_correct <- numeric(10000)

for (i in 1:10000) {
  babies <- sample(c(1, 2, 3, 4))
  num_correct[i] <- sum(babies == mothers)
}

p_all4 <- mean(num_correct == 4)
p_none <- mean(num_correct == 0)

p_all4
p_none
```

**After you run the cell:** Read <span style="color:#7c3aed">p_all4</span> and <span style="color:#7c3aed">p_none</span> carefully — small probabilities! Simulation approximates the true chance from probability theory.


### 6.4 Distribution of match counts

**Why we do this:** `table()` counts how often each outcome (0, 1, 2, 3, 4 matches) occurred.

**Your turn next (exercise notebook):** Run `table(num_correct)` after your 10,000-trial simulation.


```r
table(num_correct)
```
