### Problem 1 — Univariate: one numerical variable

Explore **flipper length** in the `palmerpenguins` dataset.

**Tasks:** histogram · dot plot · compute mean, median, SD, IQR.


### Code anatomy — Problem 1

| Part | Role |
|------|------|
| <span style="color:#2563eb">mean(x, na.rm=TRUE)</span> | Sample mean $\bar{x}$ (ignore NA) |
| <span style="color:#2563eb">median(x, na.rm=TRUE)</span> | Median |
| <span style="color:#2563eb">sd(x, na.rm=TRUE)</span> | Sample standard deviation $s$ |
| <span style="color:#2563eb">IQR(x, na.rm=TRUE)</span> | Interquartile range |
| <span style="color:#2563eb">geom_histogram()</span> | Histogram for one numerical variable |


````r
if (!requireNamespace("palmerpenguins", quietly = TRUE)) {
  install.packages("palmerpenguins", repos = "https://cloud.r-project.org")
}
library(palmerpenguins)

x <- penguins$flipper_length_mm

mean(x, na.rm = TRUE)
median(x, na.rm = TRUE)
sd(x, na.rm = TRUE)
IQR(x, na.rm = TRUE)

ggplot(penguins, aes(x = flipper_length_mm)) +
  geom_histogram(binwidth = 5, boundary = 180, fill = "steelblue", color = "white") +
  labs(title = "Penguin flipper length", x = "Flipper length (mm)", y = "Count")
````

**Read output:** <span style="color:#7c3aed">mean, median, sd, IQR</span> give the four numeric summaries. Project 3 requires one numeric summary and one graph **per variable**.


### Problem 2 — Univariate: one categorical variable

Explore **species** distribution in the penguins dataset.


````r
penguins %>%
  count(species) %>%
  mutate(prop = n / sum(n))

ggplot(penguins, aes(x = species)) +
  geom_bar(fill = "darkorange") +
  labs(title = "Penguin species", x = "Species", y = "Count")
````

### Problem 3 — Bivariate: numerical by group (boxplots)

Compare **flipper length** across three **species** (numerical × categorical).


### Code anatomy — Problem 3

| Part | Role |
|------|------|
| <span style="color:#2563eb">group_by(group)</span> | Split by categorical variable |
| <span style="color:#2563eb">summarize(...)</span> | One summary row per group |
| <span style="color:#2563eb">geom_boxplot()</span> | Side-by-side boxplots |
| <span style="color:#dc2626">species</span> | Replace with your grouping variable |
| <span style="color:#dc2626">flipper_length_mm</span> | Replace with your numerical variable |


````r
penguins %>%
  group_by(species) %>%
  summarize(
    n = n(),
    median_flip = median(flipper_length_mm, na.rm = TRUE),
    iqr_flip    = IQR(flipper_length_mm, na.rm = TRUE),
    mean_flip   = mean(flipper_length_mm, na.rm = TRUE),
    sd_flip     = sd(flipper_length_mm, na.rm = TRUE)
  )

ggplot(penguins, aes(x = species, y = flipper_length_mm, fill = species)) +
  geom_boxplot() +
  labs(title = "Flipper length by species",
       x = "Species", y = "Flipper length (mm)") +
  theme(legend.position = "none")
````

**Read output:** <span style="color:#7c3aed">median_flip, iqr_flip</span> compare center and spread per group. Boxplots show the five-number summary for each species side by side.


### Problem 4 — Bivariate: two numerical variables (scatterplot)

Explore the relationship between **bill length** and **flipper length**.


````r
penguins_clean <- penguins %>%
  filter(!is.na(bill_length_mm), !is.na(flipper_length_mm))

ggplot(penguins_clean, aes(x = bill_length_mm, y = flipper_length_mm, color = species)) +
  geom_point(alpha = 0.7) +
  labs(title = "Bill length vs Flipper length",
       x = "Bill length (mm)", y = "Flipper length (mm)")
````

**Describe:** form (linear/curved), direction (positive/negative/none), strength (strong/moderate/weak). Coloring by species reveals within-group patterns.


### Problem 5 — Bivariate: two categorical variables (stacked bar chart + two-way table)

Explore the joint distribution of **species** and **island**.


### Code anatomy — Problem 5

| Part | Role |
|------|------|
| <span style="color:#2563eb">count(var1, var2)</span> | Counts per combination of two categorical vars |
| <span style="color:#2563eb">geom_bar(position="fill")</span> | 100% stacked bar (proportions) |
| <span style="color:#2563eb">pivot_wider()</span> | Reshape count table from long to wide |
| <span style="color:#dc2626">species</span> | Replace with your first categorical variable |
| <span style="color:#dc2626">island</span> | Replace with your second categorical variable |


````r
# Two-way count table
penguins %>%
  count(species, island) %>%
  pivot_wider(names_from = island, values_from = n, values_fill = 0)

# Row proportions
penguins %>%
  count(species, island) %>%
  group_by(species) %>%
  mutate(row_prop = n / sum(n))

# 100% stacked bar
ggplot(penguins, aes(x = island, fill = species)) +
  geom_bar(position = "fill") +
  labs(title = "Species composition per island",
       x = "Island", y = "Proportion") +
  scale_y_continuous(labels = scales::percent)
````

**Read output:** <span style="color:#7c3aed">pivot_wider result</span> shows the count table. The 100% stacked bar makes within-island composition easy to compare.
