## Part 1 — Balls in a bag (without replacement)

A bag has **3 red** and **3 white** balls. Draw **2 without replacement**. Estimate $P(\text{W then R})$.


### 1.1 — Build the bag and simulate 10 draws

**Why we do this:** `rep()` creates the bag; a `for` loop with `if` records W-then-R successes.

**Your turn next (exercise notebook):** Reproduce in the exercise notebook with empty cells.
### Code anatomy — rep() + sample without replacement

| Part | Role |
|------|------|
| <span style="color:#2563eb">rep</span> | repeat labels: 3 R, 3 W |
| <span style="color:#2563eb">sample</span> | draw 2 without replacement |
| <span style="color:#2563eb">if</span> | record success only when condition holds |
| <span style="color:#059669">successes[k]</span> | store 1 for replicate k |


```r
balls <- rep(c("R", "W"), c(3, 3))
replicates <- 10
successes <- vector("numeric", replicates)
set.seed(109)

for (k in 1:replicates) {
  draw <- sample(balls, size = 2, replace = FALSE)
  if (draw[1] == "W" & draw[2] == "R") {
    successes[k] <- 1
  }
}

successes
```

### 1.5 — Many replicates

**Why we do this:** Theory: $P(W \text{ then } R) = \frac{3}{6} \times \frac{3}{5} = \frac{3}{10}$ on the 3R3W bag.


```r
balls <- rep(c("R", "W"), c(3, 3))
replicates <- 10000
successes <- vector("numeric", replicates)
set.seed(109)

for (k in 1:replicates) {
  draw <- sample(balls, size = 2, replace = FALSE)
  if (draw[1] == "W" & draw[2] == "R") {
    successes[k] <- 1
  }
}

mean(successes)
```

### 1.6 — Exactly one red (use OR)

**Why we do this:** Success when the two draws are (R,W) **or** (W,R) — use `|` inside `if`.

**Your turn next (exercise notebook):** Modify the exercise notebook `if` line yourself.


```r
balls <- rep(c("R", "W"), c(3, 3))
replicates <- 10000
successes <- vector("numeric", replicates)
set.seed(109)

for (k in 1:replicates) {
  draw <- sample(balls, size = 2, replace = FALSE)
  if ((draw[1] == "R" & draw[2] == "W") | (draw[1] == "W" & draw[2] == "R")) {
    successes[k] <- 1
  }
}

mean(successes)
```

## Part 2 — Population with conditional structure

~20% of children and ~3% of adults like gummy vitamins; 50–50 child/adult. Estimate $P(C \cap G)$ and $P(G)$.


### Part 2 — Conditional probabilities with `if`

**Why we do this:** First draw child/adult; then draw gummy preference with **different** `prob` depending on `child[i]`.
### Code anatomy — sample(..., prob = )

| Part | Role |
|------|------|
| <span style="color:#2563eb">sample</span> | random category |
| <span style="color:#dc2626">prob = c(...)</span> | weights — must sum to 1 |
| <span style="color:#2563eb">if</span> | branch on child vs adult |


```r
n <- 10000
child <- numeric(n)
gummy <- numeric(n)
set.seed(109)

for (i in 1:n) {
  child[i] <- sample(c(0, 1), 1)
  if (child[i] == 1) {
    gummy[i] <- sample(c(0, 1), 1, prob = c(0.8, 0.2))
  } else {
    gummy[i] <- sample(c(0, 1), 1, prob = c(0.97, 0.03))
  }
}

p_c_and_g <- mean(child == 1 & gummy == 1)
p_g <- mean(gummy == 1)
p_c_and_g
p_g
```

**After you run the cell:** `mean(child == 1 & gummy == 1)` estimates $P(C \cap G)$ from the simulated population.


## Part 3 — Diagnostic test

Prevalence 1%; sensitivity 95%; specificity 90%. Simulate 10,000 and estimate $P(D \mid \text{positive})$.


### Part 3 — Diagnostic test / Bayes-style simulation

**Why we do this:** Nested `if`: disease status first, then test result with different probabilities given disease.

**Your turn next (exercise notebook):** Estimate $P(D \mid +)$ with `mean(disease == 1 & test == 1) / mean(test == 1)`.


```r
n <- 10000
disease <- numeric(n)
test_result <- numeric(n)
prev <- 0.01
sensitivity <- 0.95
specificity <- 0.90
set.seed(109)

for (i in 1:n) {
  disease[i] <- sample(c(0, 1), 1, prob = c(1 - prev, prev))
  if (disease[i] == 1) {
    test_result[i] <- sample(c(0, 1), 1, prob = c(1 - sensitivity, sensitivity))
  } else {
    test_result[i] <- sample(c(0, 1), 1, prob = c(specificity, 1 - specificity))
  }
}

ppv <- mean(disease == 1 & test_result == 1) / mean(test_result == 1)
ppv
```

**After you run the cell:** `ppv` approximates $P(D \mid \text{positive})$ — small when prevalence is low even with good sensitivity.
